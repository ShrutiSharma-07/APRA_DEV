```
from snowflake.snowpark.functions import sum as sum_func, regexp_replace, upper, trim
from snowflake.snowpark.types import StringType, FloatType, ArrayType, MapType, VariantType
import json

def create_udf(session):
    """
    Creates the JavaScript UDF for composer name matching.
    
    Args:
        session: Snowflake session object
    """
    # JavaScript UDF for name matching
    create_udf_sql = """
    CREATE OR REPLACE FUNCTION MATCH_COMPOSER_NAMES(adc_parts STRING, mzk_parts STRING)
    RETURNS FLOAT
    LANGUAGE JAVASCRIPT
    AS
    $$

    function matchComposerNames(adc_parts, mzk_parts) {
        try {
            // Parse the JSON strings into arrays
            let adcComposers = JSON.parse(adc_parts);
            let mzkComposers = JSON.parse(mzk_parts);
            
            // Clean the arrays - handle escaped delimiters and empty entries
            adcComposers = cleanComposerArray(adcComposers);
            mzkComposers = cleanComposerArray(mzkComposers);
            
            // Initialize total score sum
            let totalScore = 0;
            
            // For each ADC composer, find the best matching MZK composer
            for (let i = 0; i < adcComposers.length; i++) {
                let adcComposer = adcComposers[i].toString().trim().toUpperCase();
                let bestScore = 0;
                
                // Try matching with each MZK composer
                for (let j = 0; j < mzkComposers.length; j++) {
                    let mzkComposer = mzkComposers[j].toString().trim().toUpperCase();
                    let score = 0;
                    
                    // Check for exact match
                    if (adcComposer === mzkComposer) {
                        score = 1.0;
                    } 
                    // Enhanced name part matching
                    else {
                        score = namePartsScore(adcComposer, mzkComposer);
                        
                        // If name part score is low, fall back to Jaro-Winkler
                        if (score < 0.8) {
                            score = Math.max(score, jaroWinklerSimilarity(adcComposer, mzkComposer));
                        }
                    }
                    
                    // Update best score if this match is better
                    if (score > bestScore) {
                        bestScore = score;
                    }
                }
                
                // Only add to total if the individual match score is > 0.8
                if (bestScore > 0.8) {
                    totalScore += bestScore;
                } else {
                    // Assign 0 for poor matches
                    totalScore += 0;
                }
            }
            
            // Return the sum of all filtered best scores
            return totalScore;
        } catch (e) {
            // Return error details if something goes wrong
            return 0; // Return 0 on error
        }
    }

    // Enhanced function to match composer names based on name parts with improved fuzzy matching
    function namePartsScore(name1, name2) {
        // Parse names into parts
        let parts1 = name1.split(' ').filter(part => part.length > 0);
        let parts2 = name2.split(' ').filter(part => part.length > 0);
        
        // If both have only one part or no parts, rely on direct comparison
        if ((parts1.length <= 1 && parts2.length <= 1) || parts1.length === 0 || parts2.length === 0) {
            return name1 === name2 ? 1.0 : 0.0;
        }
        
        // Check for typical name formats
        if (parts1.length >= 2 && parts2.length >= 2) {
            // Check for reversed name formats with FUZZY matching for name parts
            if (parts1.length === 2 && parts2.length === 2) {
                // Get similarity scores for parts in reverse order
                let firstLastSim = jaroWinklerSimilarity(parts1[0], parts2[1]);
                let lastFirstSim = jaroWinklerSimilarity(parts1[1], parts2[0]);
                
                // If both parts match with high similarity (>0.9), consider it a strong match
                if (firstLastSim > 0.95 && lastFirstSim > 0.9) {
                    // Scale score based on actual similarity
                    return 0.9 + ((firstLastSim + lastFirstSim) / 2) * 0.1;
                }
            }
            
            // Check for exact matches in reversed order
            if (parts1[0] === parts2[parts2.length - 1] && parts1[parts1.length - 1] === parts2[0]) {
                return 1.0;
            }
            
            // Case A: "Last First" vs "First Last"
            if (parts1[0] === parts2[1] && parts1[1] === parts2[0]) {
                return 1.0;
            }
            
            // Case B: "Last First Middle" vs "First Middle Last"
            if (parts1.length >= 3 && parts2.length >= 3) {
                if (parts1[0] === parts2[2] && parts1[1] === parts2[0]) {
                    // Check if middle parts also match or are contained
                    if (parts1.length >= 3 && parts2.length >= 2) {
                        let middleMatchScore = 0;
                        let totalMiddleParts = parts1.length - 2; // Count middle parts in name1
                        
                        // Check middle parts for partial matching
                        for (let i = 2; i < parts1.length; i++) {
                            let bestMiddleMatch = 0;
                            for (let j = 1; j < parts2.length - 1; j++) {
                                if (parts1[i] === parts2[j] || parts2[j].includes(parts1[i]) || parts1[i].includes(parts2[j])) {
                                    bestMiddleMatch = 1.0;
                                    break;
                                }
                            }
                            middleMatchScore += bestMiddleMatch;
                        }
                        
                        // If ALL middle parts match (full matching), return 1.0 instead of 0.9
                        if (middleMatchScore === totalMiddleParts) {
                            return 1.0;  // Perfect match for all components, just in different order
                        }
                        // If some middle parts match but not all, return 0.9
                        else if (middleMatchScore > 0) {
                            return 0.9;
                        }
                    }
                    
                    // First and last parts match, even if middle parts don't fully match
                    return 0.85;
                }
            }
            
            // Case C: "Last First Middle" vs "First Last"
            if (parts1.length >= 3 && parts1[0] === parts2[1] && parts1[2] === parts2[0]) {
                return 0.9;
            }
            
            // Case D: "Last First" vs "First Middle Last"
            if (parts1.length >= 2 && parts2.length >= 3 && parts1[0] === parts2[2] && parts1[1] === parts2[0]) {
                return 0.9;
            }
            
            // ENHANCED MATCHING: Check for fuzzy part matches
            let bestPartMatches = [];
            
            // For each part in name1, find its best matching part in name2
            for (let i = 0; i < parts1.length; i++) {
                let bestMatch = { score: 0, index: -1 };
                
                for (let j = 0; j < parts2.length; j++) {
                    let similarity = jaroWinklerSimilarity(parts1[i], parts2[j]);
                    
                    // Consider high similarities as potential matches
                    if (similarity > 0.85 && similarity > bestMatch.score) {
                        bestMatch.score = similarity;
                        bestMatch.index = j;
                    }
                }
                
                if (bestMatch.index >= 0) {
                    bestPartMatches.push({
                        part1Index: i,
                        part2Index: bestMatch.index,
                        score: bestMatch.score
                    });
                }
            }
            
            // Calculate a fuzzy match score based on the best part matches
            if (bestPartMatches.length > 0) {
                let totalScore = 0;
                for (let i = 0; i < bestPartMatches.length; i++) {
                    totalScore += bestPartMatches[i].score;
                }
                let averageScore = totalScore / parts1.length;
                
                // Adjust final score to reward high similarity matches
                return 0.7 + (averageScore * 0.3);
            }
            
            // Check for scrambled parts with at least one exact match
            let matchingParts = 0;
            let matchingChars = 0;
            
            // Count exactly matching parts
            for (let i = 0; i < parts1.length; i++) {
                if (parts2.includes(parts1[i])) {
                    matchingParts++;
                    matchingChars += parts1[i].length;
                }
            }
            
            // If we have a decent number of matching parts, consider it a match
            if (matchingParts >= 2 || (matchingParts === 1 && parts1.length <= 3 && parts2.length <= 3)) {
                // Calculate score based on proportion of matching characters to total characters
                let totalChars1 = parts1.join('').length;
                let charMatchScore = matchingChars / totalChars1;
                
                // Adjusted score based on matching parts
                return 0.7 + (charMatchScore * 0.3);
            }
        }
        
        // Default case: fall back to Jaro-Winkler similarity
        return jaroWinklerSimilarity(name1, name2);
    }

    // Helper function to clean composer arrays
    function cleanComposerArray(composers) {
        // Filter out entries that contain escaped delimiters (like "#\\")
        return composers.filter(composer => {
            let str = composer.toString().trim();
            
            // Check if this entry is just an escaped delimiter
            if (/^[^#]+#\\+$/.test(str)) {
                return false;
            }
            
            // Keep entries that have actual content
            return str.length > 0;
        });
    }

    // Implementation of Jaro-Winkler similarity function
    function jaroWinklerSimilarity(s1, s2) {
        // If either string is empty, return 0
        if (!s1 || !s2 || s1.length === 0 || s2.length === 0) {
            return 0;
        }
        
        // If both strings are identical, return 1
        if (s1 === s2) {
            return 1;
        }
        
        // Calculate Jaro Distance
        let m = 0; // matching characters
        let t = 0; // transpositions
        let range = Math.floor(Math.max(s1.length, s2.length) / 2) - 1;
        range = Math.max(0, range); // Ensure range is not negative
        
        let s1Matches = new Array(s1.length).fill(false);
        let s2Matches = new Array(s2.length).fill(false);
        
        // Find matching characters within range
        for (let i = 0; i < s1.length; i++) {
            let start = Math.max(0, i - range);
            let end = Math.min(i + range + 1, s2.length);
            
            for (let j = start; j < end; j++) {
                if (!s2Matches[j] && s1[i] === s2[j]) {
                    s1Matches[i] = true;
                    s2Matches[j] = true;
                    m++;
                    break;
                }
            }
        }
        
        // If no matching characters, return 0
        if (m === 0) {
            return 0;
        }
        
        // Count transpositions
        let k = 0;
        for (let i = 0; i < s1.length; i++) {
            if (s1Matches[i]) {
                while (k < s2.length && !s2Matches[k]) {
                    k++;
                }
                
                if (k < s2.length && s1[i] !== s2[k]) {
                    t++;
                }
                
                if (k < s2.length) {
                    k++;
                }
            }
        }
        
        // Calculate Jaro similarity
        t = Math.floor(t / 2);
        let jaroSim = (m / s1.length + m / s2.length + (m - t) / m) / 3;
        
        // Calculate Jaro-Winkler similarity
        let p = 0.1; // scaling factor
        let l = 0;   // length of common prefix (max 4)
        
        // Count common prefix up to 4 characters
        for (let i = 0; i < Math.min(4, Math.min(s1.length, s2.length)); i++) {
            if (s1[i] === s2[i]) {
                l++;
            } else {
                break;
            }
        }
        
        // Apply Winkler modification
        return jaroSim + (l * p * (1 - jaroSim));
    }

    return matchComposerNames(ADC_PARTS, MZK_PARTS);
    $$;
    """

    session.sql(create_udf_sql).collect()
    
def prepare_test_data(session):
    """
    Creates and returns test data for composer name matching.
    
    Args:
        session: Snowflake session object
        
    Returns:
        tuple: (adc_test_data_df, mzk_test_data_df) - DataFrames with test data
    """
    # APRA Data
    adc_composers_data = [
        {"ADC_COMPOSER_ID": 43774657, "APRA_WORK_ID": "GW01421940", "IPI": "323735", "COMPOSER_NAME": "AHMADZAI MIRWAIS", "ISWC": "T0700428041", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c1"},
        {"ADC_COMPOSER_ID": 43774658, "APRA_WORK_ID": "GW01421941", "IPI": "523736", "COMPOSER_NAME": "SMITH JOHN", "ISWC": "T0700428042", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c2"},
        {"ADC_COMPOSER_ID": 43774659, "APRA_WORK_ID": "GW01421942", "IPI": "723737", "COMPOSER_NAME": "WILSON BRIAN", "ISWC": "T0700428043", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c3"},
        {"ADC_COMPOSER_ID": 43774660, "APRA_WORK_ID": "GW01421943", "IPI": "823738", "COMPOSER_NAME": "TAYLOR D", "ISWC": "T0700428044", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c4"},
        {"ADC_COMPOSER_ID": 43774661, "APRA_WORK_ID": "GW01421944", "IPI": "923739", "COMPOSER_NAME": "J WAGNER", "ISWC": "T0700428045", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c5"},
        {"ADC_COMPOSER_ID": 43774662, "APRA_WORK_ID": "GW01421945", "IPI": "123740", "COMPOSER_NAME": "BEETHOVEN LUDWIG VAN", "ISWC": "T0700428046", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c6"},
        {"ADC_COMPOSER_ID": 43774663, "APRA_WORK_ID": "GW01421946", "IPI": "223741", "COMPOSER_NAME": "ARMSTRONG LOUIS DANIEL", "ISWC": "T0700428047", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c7"},
        {"ADC_COMPOSER_ID": 43774664, "APRA_WORK_ID": "GW01421947", "IPI": "323742", "COMPOSER_NAME": "LENNON JOHN WINSTON", "ISWC": "T0700428048", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c8"},
        {"ADC_COMPOSER_ID": 44100861, "APRA_WORK_ID": "GW02005365", "IPI": "445678", "COMPOSER_NAME": "MARTIN HUGH", "ISWC": "T0700726620", "MUZOOKA_TRACK_ID": "mz-2bfcf0aa02a2eb187fb02952d46203c8"}
    ]
    
    # Muzooka Data
    muzooka_composers_data = [
        {"COMPOSER_ID": "mz-comp-1", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c1", "COMPOSER_NAME": "MIRUAIS AHMADZAI", "IPI": "323735", "ISWC": "T0700428041", "APRA_WORK_ID": "GW01421940"},
        {"COMPOSER_ID": "mz-comp-2", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c2", "COMPOSER_NAME": "JOHN SMITH", "IPI": "523736", "ISWC": "T0700428042", "APRA_WORK_ID": "GW01421941"},
        {"COMPOSER_ID": "mz-comp-3", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c3", "COMPOSER_NAME": "BRIAN WILSON", "IPI": "723737", "ISWC": "T0700428043", "APRA_WORK_ID": "GW01421942"},
        {"COMPOSER_ID": "mz-comp-4", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c4", "COMPOSER_NAME": "DAVID TAYLOR", "IPI": "823738", "ISWC": "T0700428044", "APRA_WORK_ID": "GW01421943"},
        {"COMPOSER_ID": "mz-comp-5", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c5", "COMPOSER_NAME": "JACK WAGNER", "IPI": "923739", "ISWC": "T0700428045", "APRA_WORK_ID": "GW01421944"},
        {"COMPOSER_ID": "mz-comp-6", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c6", "COMPOSER_NAME": "LUDWIG VAN BEETHOVEN", "IPI": "123740", "ISWC": "T0700428046", "APRA_WORK_ID": "GW01421945"},
        {"COMPOSER_ID": "mz-comp-7", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c7", "COMPOSER_NAME": "LOUIS D. ARMSTRONG", "IPI": "223741", "ISWC": "T0700428047", "APRA_WORK_ID": "GW01421946"},
        {"COMPOSER_ID": "mz-comp-8", "MUZOOKA_TRACK_ID": "mz-fdbe5c86458643076e5514de8111d3c8", "COMPOSER_NAME": "JOHN LENNON", "IPI": "323742", "ISWC": "T0700428048", "APRA_WORK_ID": "GW01421947"},
        {"COMPOSER_ID": "mz-comp-9", "MUZOOKA_TRACK_ID": "mz-2bfcf0aa02a2eb187fb02952d46203c8", "COMPOSER_NAME": "HUGH MARTIN", "IPI": "445678", "ISWC": "T0700726620", "APRA_WORK_ID": "GW02005365"}
    ]
    
    # Create DataFrames from the test data
    adc_test_df = session.create_dataframe(adc_composers_data)
    mzk_test_df = session.create_dataframe(muzooka_composers_data)
    
    # Process the test data to format it properly
    adc_test_processed = adc_test_df.with_column("COMPOSER_NAME", upper(trim("COMPOSER_NAME")))
    mzk_test_processed = mzk_test_df.with_column("COMPOSER_NAME", upper(trim("COMPOSER_NAME")))
    
    # Use manual SQL approach for processing composer names
    adc_test_df.create_or_replace_temp_view("TEMP_ADC_VIEW")
    mzk_test_df.create_or_replace_temp_view("TEMP_MZK_VIEW")
    
    # Process ADC test data with SQL
    adc_sql = """
    SELECT 
        ADC_COMPOSER_ID,
        APRA_WORK_ID,
        IPI,
        UPPER(TRIM(COMPOSER_NAME)) AS APRA_COMPOSER_NAME,
        ISWC,
        MUZOOKA_TRACK_ID,
        PARSE_JSON(TO_JSON(SPLIT(UPPER(TRIM(COMPOSER_NAME)), ' '))) AS APRA_DELIMITED_PARTS,
        ARRAY_SIZE(SPLIT(UPPER(TRIM(COMPOSER_NAME)), ' ')) AS APRA_COMPOSER_CT
    FROM 
        TEMP_ADC_VIEW
    """
    
    adc_test_processed = session.sql(adc_sql)
    
    # Process MZK test data with SQL
    mzk_sql = """
    SELECT 
        COMPOSER_ID AS MZK_COMPOSER_ID,
        MUZOOKA_TRACK_ID,
        UPPER(TRIM(COMPOSER_NAME)) AS MZK_COMPOSER_NAME,
        IPI AS MZK_IPI,
        ISWC AS MZK_ISWC,
        APRA_WORK_ID,
        PARSE_JSON(TO_JSON(SPLIT(UPPER(TRIM(COMPOSER_NAME)), ' '))) AS MZK_DELIMITED_PARTS,
        ARRAY_SIZE(SPLIT(UPPER(TRIM(COMPOSER_NAME)), ' ')) AS MZK_COMPOSER_CT
    FROM 
        TEMP_MZK_VIEW
    """
    
    mzk_test_processed = session.sql(mzk_sql)
    
    return adc_test_processed, mzk_test_processed


def run_test_mode(session):
    """
    Runs the name matching test using sample data
    
    Args:
        session: Snowflake session object
    """
    print("Running composer name matching test with sample data...")
    
    # Get test data
    adc_test_df, mzk_test_df = prepare_test_data(session)
    
    # Create temporary views
    adc_test_df.create_or_replace_temp_view("ADC_COMPOSER_TEST_VIEW")
    mzk_test_df.create_or_replace_temp_view("MZK_COMPOSER_TEST_VIEW")
    
    # Run the matching test
    test_match_sql = """
    SELECT
        a.MUZOOKA_TRACK_ID,
        a.ADC_COMPOSER_ID,
        a.APRA_COMPOSER_NAME,
        a.IPI AS APRA_IPI,
        m.MZK_COMPOSER_ID,
        m.MZK_COMPOSER_NAME,
        m.MZK_IPI,
        a.APRA_DELIMITED_PARTS,
        a.APRA_COMPOSER_CT,
        m.MZK_DELIMITED_PARTS,
        m.MZK_COMPOSER_CT,
        MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS) as COMPOSER_MATCH_SCORE,
        (MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS)/a.APRA_COMPOSER_CT) as NORMALIZED_SCORE
    FROM
        ADC_COMPOSER_TEST_VIEW a
    JOIN 
        MZK_COMPOSER_TEST_VIEW m
    ON 
        a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    ORDER BY
        NORMALIZED_SCORE DESC
    """
    
    # Execute matching
    test_results = session.sql(test_match_sql)
    
    # Show test results
    print("\n=== TEST RESULTS ===")
    test_results.show()
    
    # Calculate summary statistics
    summary_sql = """
    WITH test_results AS (
        SELECT
            a.MUZOOKA_TRACK_ID,
            a.ADC_COMPOSER_ID,
            a.APRA_COMPOSER_NAME,
            m.MZK_COMPOSER_NAME,
            MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS) as COMPOSER_MATCH_SCORE,
            (MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS)/a.APRA_COMPOSER_CT) as NORMALIZED_SCORE
        FROM
            ADC_COMPOSER_TEST_VIEW a
        JOIN 
            MZK_COMPOSER_TEST_VIEW m
        ON 
            a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    )
    SELECT
        COUNT(CASE WHEN NORMALIZED_SCORE = 1.0 THEN 1 END) AS EXACT_MATCH_COUNT,
        COUNT(CASE WHEN NORMALIZED_SCORE < 1.0 AND NORMALIZED_SCORE >= 0.9 THEN 1 END) AS HIGH_MATCH_COUNT,
        COUNT(CASE WHEN NORMALIZED_SCORE < 0.9 AND NORMALIZED_SCORE >= 0.8 THEN 1 END) AS MEDIUM_MATCH_COUNT,
        COUNT(*) AS TOTAL_MATCHES
    FROM
        test_results
    """
    
    summary_results = session.sql(summary_sql)
    
    print("\n=== MATCH SUMMARY ===")
    summary_results.show()
    
    # Print detailed example for each match type
    detailed_examples_sql = """
    WITH test_results AS (
        SELECT
            a.MUZOOKA_TRACK_ID,
            a.ADC_COMPOSER_ID,
            a.APRA_COMPOSER_NAME,
            m.MZK_COMPOSER_NAME,
            MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS) as COMPOSER_MATCH_SCORE,
            (MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS)/a.APRA_COMPOSER_CT) as NORMALIZED_SCORE
        FROM
            ADC_COMPOSER_TEST_VIEW a
        JOIN 
            MZK_COMPOSER_TEST_VIEW m
        ON 
            a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    )
    SELECT 
        CASE 
            WHEN NORMALIZED_SCORE = 1.0 THEN 'EXACT MATCH'
            WHEN NORMALIZED_SCORE >= 0.9 THEN 'HIGH MATCH'
            WHEN NORMALIZED_SCORE >= 0.8 THEN 'MEDIUM MATCH'
            ELSE 'LOW MATCH'
        END as MATCH_TYPE,
        APRA_COMPOSER_NAME,
        MZK_COMPOSER_NAME,
        NORMALIZED_SCORE
    FROM
        test_results
    ORDER BY
        NORMALIZED_SCORE DESC
    """
    
    examples_results = session.sql(detailed_examples_sql)
    
    print("\n=== MATCH EXAMPLES ===")
    examples_results.show()
    
    print("Test execution complete!")


def run_full_dataset(session):
    """
    Runs the name matching process on the full dataset
    
    Args:
        session: Snowflake session object
    """
    print("Processing full dataset for composer name matching...")
    
    # Load the data from the composer tables
    adc_df = session.table("ADC_COMPOSER_DELIMITER_COUNT")
    mzk_df = session.table("MZK_COMPOSER_DELIMITER_COUNT")

    # Create temporary views to make SQL operations easier
    adc_df.create_or_replace_temp_view("ADC_COMPOSER_TEMP_VIEW")
    mzk_df.create_or_replace_temp_view("MZK_COMPOSER_TEMP_VIEW")

    # Main matching SQL
    # Main matching SQL (continued)
    match_sql = """
    SELECT DISTINCT
        a.ADC_COMPOSER_ID,  
        a.APRA_COMPOSER_NAME,
        a.MUZOOKA_TRACK_ID,
        --m.MZK_COMPOSER_ID,
        UPPER(m.MZK_COMPOSER_NAME) AS MZK_COMPOSER_NAME,
        a.APRA_DELIMITED_PARTS,
        a.APRA_COMPOSER_CT,  
        m.MZK_DELIMITED_PARTS,
        m.MZK_COMPOSER_CT,  
        --MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS) as VARIANT_MATCH_SCORE,  
        (MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS)/a.APRA_COMPOSER_CT) as COMPOSER_MATCH_SCORE  
    FROM 
        ADC_COMPOSER_TEMP_VIEW a
    INNER JOIN 
        MZK_COMPOSER_TEMP_VIEW m
    ON 
        a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    WHERE
        MATCH_COMPOSER_NAMES(a.APRA_DELIMITED_PARTS, m.MZK_DELIMITED_PARTS) >= 0.8 AND COMPOSER_MATCH_SCORE >= 0.8  
    """

    # Execute the SQL query
    matched_df = session.sql(match_sql)

    # Create or replace the output table
    matched_df.write.mode("overwrite").save_as_table("COMPOSER_NAME_MATCH_RESULTS")

    # Show the results
    print("\n=== MATCHING RESULTS SAMPLE ===")
    matched_df.show()

    # Summary SQL to get match statistics
    summary_sql = """
    SELECT
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE = 1.0 THEN 1 END) AS EXACT_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 1.0 AND COMPOSER_MATCH_SCORE >= 0.9 THEN 1 END) AS HIGH_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 0.9 AND COMPOSER_MATCH_SCORE >= 0.8 THEN 1 END) AS MEDIUM_MATCH_COUNT,
        COUNT(*) AS TOTAL_MATCHES
    FROM
        COMPOSER_NAME_MATCH_RESULTS
    """

    # Debug SQL to show score distribution
    debug_sql = """
    SELECT 
        ROUND(COMPOSER_MATCH_SCORE, 2) as SCORE_RANGE,
        COUNT(*) as COUNT
    FROM 
        COMPOSER_NAME_MATCH_RESULTS
    GROUP BY 
        ROUND(COMPOSER_MATCH_SCORE, 2)
    ORDER BY 
        SCORE_RANGE DESC
    """

    # Show debug information
    debug_df = session.sql(debug_sql)
    print("\n=== SCORE DISTRIBUTION ===")
    debug_df.show()

    # Show summary information
    summary_df = session.sql(summary_sql)
    print("\n=== MATCH SUMMARY ===")
    summary_df.show()

    print("Processing complete. Results stored in COMPOSER_NAME_MATCH_RESULTS table.")


def main(session):
    
    run_mode = 'full'  # Options: 'full' or 'test'
    
    # Create the JavaScript UDF
    create_udf(session)
    
    if run_mode.lower() == 'test':
        run_test_mode(session)
    elif run_mode.lower() == 'full':
        run_full_dataset(session)
    else:
        print(f"Invalid run mode: {run_mode}. Please specify 'test' or 'full'")


if __name__ == "__main__":
    # Call the main function
    main(session)
```
