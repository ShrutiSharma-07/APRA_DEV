```
#MATCHING SCRIPT


from snowflake.snowpark.functions import sum as sum_func, regexp_replace, upper, trim, col, when
from snowflake.snowpark.types import StringType, FloatType, ArrayType, MapType, VariantType
import json

def create_udf(session):
    """Create JavaScript UDF for name matching using Jaro-Winkler similarity"""
    
    create_udf_sql = """
    CREATE OR REPLACE FUNCTION MATCH_COMPOSER_NAMES_JW(name1 STRING, name2 STRING)
    RETURNS FLOAT
    LANGUAGE JAVASCRIPT
    AS
    $$
    
    function matchComposerNames(name1, name2) {
        try {
            // Handle null or empty inputs
            if (!name1 || !name2 || name1.length === 0 || name2.length === 0) {
                return 0.0;
            }
            
            // Clean and normalize the names
            let cleanName1 = name1.toString().trim().toUpperCase();
            let cleanName2 = name2.toString().trim().toUpperCase();
            
            // If names are identical after cleaning, return perfect score
            if (cleanName1 === cleanName2) {
                return 1.0;
            }
            
            // Calculate Jaro-Winkler similarity
            return jaroWinklerSimilarity(cleanName1, cleanName2);
            
        } catch (e) {
            // Return 0 on error
            return 0.0;
        }
    }

    // Implementation of Jaro-Winkler similarity function
    function jaroWinklerSimilarity(s1, s2) {
        // If either string is empty, return 0
        if (!s1 || !s2 || s1.length === 0 || s2.length === 0) {
            return 0;
        }
        
        // If both strings are identical, return 1
        if (s1 === s2) {
            return 1;
        }
        
        // Calculate Jaro Distance
        let m = 0; // matching characters
        let t = 0; // transpositions
        let range = Math.floor(Math.max(s1.length, s2.length) / 2) - 1;
        range = Math.max(0, range); // Ensure range is not negative
        
        let s1Matches = new Array(s1.length).fill(false);
        let s2Matches = new Array(s2.length).fill(false);
        
        // Find matching characters within range
        for (let i = 0; i < s1.length; i++) {
            let start = Math.max(0, i - range);
            let end = Math.min(i + range + 1, s2.length);
            
            for (let j = start; j < end; j++) {
                if (!s2Matches[j] && s1[i] === s2[j]) {
                    s1Matches[i] = true;
                    s2Matches[j] = true;
                    m++;
                    break;
                }
            }
        }
        
        // If no matching characters, return 0
        if (m === 0) {
            return 0;
        }
        
        // Count transpositions
        let k = 0;
        for (let i = 0; i < s1.length; i++) {
            if (s1Matches[i]) {
                while (k < s2.length && !s2Matches[k]) {
                    k++;
                }
                
                if (k < s2.length && s1[i] !== s2[k]) {
                    t++;
                }
                
                if (k < s2.length) {
                    k++;
                }
            }
        }
        
        // Calculate Jaro similarity
        t = Math.floor(t / 2);
        let jaroSim = (m / s1.length + m / s2.length + (m - t) / m) / 3;
        
        // Calculate Jaro-Winkler similarity
        let p = 0.1; // scaling factor
        let l = 0;   // length of common prefix (max 4)
        
        // Count common prefix up to 4 characters
        for (let i = 0; i < Math.min(4, Math.min(s1.length, s2.length)); i++) {
            if (s1[i] === s2[i]) {
                l++;
            } else {
                break;
            }
        }
        
        // Apply Winkler modification
        return jaroSim + (l * p * (1 - jaroSim));
    }

    return matchComposerNames(NAME1, NAME2);
    $$;
    """

    session.sql(create_udf_sql).collect()
    print("JavaScript UDF for Jaro-Winkler similarity created successfully.")


def prepare_example_data(session):
    """Prepare small example dataset for testing"""
    
    # ADC Example Data
    adc_composers_data = [
        {"ADC_COMPOSER_ID": 1, "APRA_WORK_ID": "GW01421940", "APRA_COMPOSER_NAME": "AHMADZAI MIRWAIS", "IPI": "323735", "APRA_COMPOSER_CT": 2, "MUZOOKA_TRACK_ID": "mz-track-1"},
        {"ADC_COMPOSER_ID": 2, "APRA_WORK_ID": "GW01421941", "APRA_COMPOSER_NAME": "SMITH JOHN", "IPI": "523736", "APRA_COMPOSER_CT": 2, "MUZOOKA_TRACK_ID": "mz-track-2"},
        {"ADC_COMPOSER_ID": 3, "APRA_WORK_ID": "GW01421942", "APRA_COMPOSER_NAME": "WILSON BRIAN", "IPI": "723737", "APRA_COMPOSER_CT": 2, "MUZOOKA_TRACK_ID": "mz-track-3"},
        {"ADC_COMPOSER_ID": 4, "APRA_WORK_ID": "GW01421943", "APRA_COMPOSER_NAME": "TAYLOR D", "IPI": "823738", "APRA_COMPOSER_CT": 2, "MUZOOKA_TRACK_ID": "mz-track-4"},
        {"ADC_COMPOSER_ID": 5, "APRA_WORK_ID": "GW01421944", "APRA_COMPOSER_NAME": "BEETHOVEN LUDWIG VAN", "IPI": "123740", "APRA_COMPOSER_CT": 3, "MUZOOKA_TRACK_ID": "mz-track-5"}
    ]
    
    # MZK Example Data
    muzooka_composers_data = [
        {"COMPOSER_ID": 101, "MUZOOKA_TRACK_ID": "mz-track-1", "COMPOSER_NAME": "MIRUAIS AHMADZAI", "IPI": "323735", "MZK_COMPOSER_CT": 2},
        {"COMPOSER_ID": 102, "MUZOOKA_TRACK_ID": "mz-track-2", "COMPOSER_NAME": "JOHN SMITH", "IPI": "523736", "MZK_COMPOSER_CT": 2},
        {"COMPOSER_ID": 103, "MUZOOKA_TRACK_ID": "mz-track-3", "COMPOSER_NAME": "BRIAN WILSON", "IPI": "723737", "MZK_COMPOSER_CT": 2},
        {"COMPOSER_ID": 104, "MUZOOKA_TRACK_ID": "mz-track-4", "COMPOSER_NAME": "DAVID TAYLOR", "IPI": "999999", "MZK_COMPOSER_CT": 2},  # Different IPI
        {"COMPOSER_ID": 105, "MUZOOKA_TRACK_ID": "mz-track-5", "COMPOSER_NAME": "LUDWIG VAN BEETHOVEN", "IPI": "123740", "MZK_COMPOSER_CT": 3}
    ]
    
    # Create DataFrames from the example data
    adc_example_df = session.create_dataframe(adc_composers_data)
    mzk_example_df = session.create_dataframe(muzooka_composers_data)
    
    return adc_example_df, mzk_example_df


def run_example_mode(session):
    """Run composer name matching with small example dataset"""
    
    print("Running composer name matching with example data...")
    
    # Get example data
    adc_example_df, mzk_example_df = prepare_example_data(session)
    
    # Create temporary views
    adc_example_df.create_or_replace_temp_view("ADC_COMPOSER_EXAMPLE_VIEW")
    mzk_example_df.create_or_replace_temp_view("MZK_COMPOSER_EXAMPLE_VIEW")
    
    # Run the matching using INNER JOIN on muzooka_track_id
    example_match_sql = """
    SELECT
        a.ADC_COMPOSER_ID,
        a.APRA_WORK_ID,
        a.APRA_COMPOSER_NAME,
        a.IPI AS ADC_IPI,
        a.APRA_COMPOSER_CT,
        a.MUZOOKA_TRACK_ID,
        m.MZK_COMPOSER_CT,
        m.COMPOSER_NAME,
        m.IPI AS MZK_IPI,
        CASE 
            WHEN a.IPI = m.IPI THEN 'Y' 
            ELSE 'N' 
        END AS YN_IPI_MATCH,
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) AS COMPOSER_MATCH_SCORE
    FROM
        ADC_COMPOSER_EXAMPLE_VIEW a
    INNER JOIN 
        MZK_COMPOSER_EXAMPLE_VIEW m
    ON a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    WHERE
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) >= 0.6
    ORDER BY
        a.ADC_COMPOSER_ID, COMPOSER_MATCH_SCORE DESC
    """
    
    # Execute matching
    example_results = session.sql(example_match_sql)
    
    # Show example results
    print("\n=== EXAMPLE RESULTS (Threshold >= 0.6) ===")
    example_results.show()
    
    # Calculate summary statistics
    summary_sql = """
    WITH example_results AS (
        SELECT
            a.ADC_COMPOSER_ID,
            a.APRA_COMPOSER_NAME,
            m.COMPOSER_NAME,
            CASE WHEN a.IPI = m.IPI THEN 'Y' ELSE 'N' END AS YN_IPI_MATCH,
            MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) AS COMPOSER_MATCH_SCORE
        FROM
            ADC_COMPOSER_EXAMPLE_VIEW a
        INNER JOIN 
            MZK_COMPOSER_EXAMPLE_VIEW m
        ON a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
        WHERE
            MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) >= 0.6
    )
    SELECT
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE = 1.0 THEN 1 END) AS EXACT_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 1.0 AND COMPOSER_MATCH_SCORE >= 0.9 THEN 1 END) AS HIGH_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 0.9 AND COMPOSER_MATCH_SCORE >= 0.8 THEN 1 END) AS MEDIUM_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 0.8 AND COMPOSER_MATCH_SCORE >= 0.6 THEN 1 END) AS LOW_MATCH_COUNT,
        COUNT(CASE WHEN YN_IPI_MATCH = 'Y' THEN 1 END) AS IPI_MATCH_COUNT,
        COUNT(*) AS TOTAL_MATCHES
    FROM
        example_results
    """
    
    summary_results = session.sql(summary_sql)
    print("\n=== EXAMPLE SUMMARY ===")
    summary_results.show()
    
    print("Example mode execution complete!")


def run_subset_mode(session):
    """Run composer name matching on a subset of the full dataset"""
    
    print("Running composer name matching on subset of data...")
    
    # Load subset of data (limit to reduce processing time)
    subset_sql = """
    WITH adc_subset AS (
        SELECT *
        FROM ADC_COMPOSER_DELIMITER_COUNT
        WHERE MUZOOKA_TRACK_ID IS NOT NULL
        LIMIT 1000
    ),
    mzk_subset AS (
        SELECT *
        FROM MZK_COMPOSER_DELIMITER_COUNT
        WHERE MUZOOKA_TRACK_ID IS NOT NULL
        LIMIT 1000
    )
    SELECT
        a.ADC_COMPOSER_ID,
        a.APRA_WORK_ID,
        a.APRA_COMPOSER_NAME,
        a.IPI AS ADC_IPI,
        a.APRA_COMPOSER_CT,
        a.MUZOOKA_TRACK_ID,
        m.MZK_COMPOSER_CT,
        m.COMPOSER_NAME,
        m.IPI AS MZK_IPI,
        CASE 
            WHEN a.IPI = m.IPI THEN 'Y' 
            ELSE 'N' 
        END AS YN_IPI_MATCH,
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) AS COMPOSER_MATCH_SCORE
    FROM 
        adc_subset a
    INNER JOIN 
        mzk_subset m
    ON a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    WHERE
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) >= 0.8
    ORDER BY
        a.ADC_COMPOSER_ID, COMPOSER_MATCH_SCORE DESC
    """
    
    # Execute the subset query
    subset_results = session.sql(subset_sql)
    
    # Save subset results
    subset_results.write.mode("overwrite").save_as_table("COMPOSER_NAME_MATCH_SUBSET_RESULTS")
    
    print("\n=== SUBSET MATCHING RESULTS SAMPLE ===")
    subset_results.limit(20).show()
    
    # Summary statistics for subset
    summary_sql = """
    SELECT
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE = 1.0 THEN 1 END) AS EXACT_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 1.0 AND COMPOSER_MATCH_SCORE >= 0.9 THEN 1 END) AS HIGH_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 0.9 AND COMPOSER_MATCH_SCORE >= 0.8 THEN 1 END) AS MEDIUM_MATCH_COUNT,
        COUNT(CASE WHEN YN_IPI_MATCH = 'Y' THEN 1 END) AS IPI_MATCH_COUNT,
        COUNT(*) AS TOTAL_MATCHES,
        COUNT(DISTINCT MUZOOKA_TRACK_ID) AS UNIQUE_TRACKS
    FROM
        COMPOSER_NAME_MATCH_SUBSET_RESULTS
    """
    
    summary_df = session.sql(summary_sql)
    print("\n=== SUBSET MATCH SUMMARY ===")
    summary_df.show()
    
    print("Subset mode processing complete. Results stored in COMPOSER_NAME_MATCH_SUBSET_RESULTS table.")


def run_full_mode(session):
    """Run composer name matching on the full dataset"""
    
    print("Running composer name matching on full dataset...")
    
    # Main matching SQL with INNER JOIN on muzooka_track_id
    full_match_sql = """
    SELECT
        a.ADC_COMPOSER_ID,
        a.APRA_WORK_ID,
        a.APRA_COMPOSER_NAME,
        a.IPI AS ADC_IPI,
        a.APRA_COMPOSER_CT,
        a.MUZOOKA_TRACK_ID,
        m.MZK_COMPOSER_CT,
        m.COMPOSER_NAME,
        m.IPI AS MZK_IPI,
        CASE 
            WHEN a.IPI = m.IPI THEN 'Y' 
            ELSE 'N' 
        END AS YN_IPI_MATCH,
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) AS COMPOSER_MATCH_SCORE
    FROM 
        ADC_COMPOSER_DELIMITER_COUNT a
    INNER JOIN 
        MZK_COMPOSER_DELIMITER_COUNT m
    ON a.MUZOOKA_TRACK_ID = m.MUZOOKA_TRACK_ID
    WHERE
        MATCH_COMPOSER_NAMES_JW(a.APRA_COMPOSER_NAME, m.COMPOSER_NAME) >= 0.8
    ORDER BY
        a.ADC_COMPOSER_ID, COMPOSER_MATCH_SCORE DESC
    """

    # Execute the full dataset query
    full_results = session.sql(full_match_sql)

    # Save full results
    full_results.write.mode("overwrite").save_as_table("COMPOSER_NAME_MATCH_FULL_RESULTS")

    # Show sample results
    print("\n=== FULL MATCHING RESULTS SAMPLE ===")
    full_results.limit(20).show()

    # Summary statistics for full dataset
    summary_sql = """
    SELECT
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE = 1.0 THEN 1 END) AS EXACT_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 1.0 AND COMPOSER_MATCH_SCORE >= 0.9 THEN 1 END) AS HIGH_MATCH_COUNT,
        COUNT(CASE WHEN COMPOSER_MATCH_SCORE < 0.9 AND COMPOSER_MATCH_SCORE >= 0.8 THEN 1 END) AS MEDIUM_MATCH_COUNT,
        COUNT(CASE WHEN YN_IPI_MATCH = 'Y' THEN 1 END) AS IPI_MATCH_COUNT,
        COUNT(*) AS TOTAL_MATCHES,
        COUNT(DISTINCT MUZOOKA_TRACK_ID) AS UNIQUE_TRACKS
    FROM
        COMPOSER_NAME_MATCH_FULL_RESULTS
    """

    summary_df = session.sql(summary_sql)
    print("\n=== FULL MATCH SUMMARY ===")
    summary_df.show()

    # Show score distribution
    score_distribution_sql = """
    SELECT 
        ROUND(COMPOSER_MATCH_SCORE, 1) as SCORE_RANGE,
        COUNT(*) as COUNT
    FROM 
        COMPOSER_NAME_MATCH_FULL_RESULTS
    GROUP BY 
        ROUND(COMPOSER_MATCH_SCORE, 1)
    ORDER BY 
        SCORE_RANGE DESC
    """

    score_df = session.sql(score_distribution_sql)
    print("\n=== SCORE DISTRIBUTION ===")
    score_df.show()

    print("Full mode processing complete. Results stored in COMPOSER_NAME_MATCH_FULL_RESULTS table.")


def main(session):
    """Main function to run composer name matching"""
    
    # Set run mode: 'example', 'subset', or 'full'
    run_mode = 'full'  # Change this to run different modes
    
    # Create the JavaScript UDF for Jaro-Winkler similarity
    create_udf(session)
    
    if run_mode.lower() == 'example':
        run_example_mode(session)
    elif run_mode.lower() == 'subset':
        run_subset_mode(session)
    elif run_mode.lower() == 'full':
        run_full_mode(session)
    else:
        print(f"Invalid run mode: {run_mode}. Please specify 'example', 'subset', or 'full'")
        return
    
    print(f"\nComposer name matching completed in '{run_mode}' mode.")
    print("All matching uses INNER JOIN on muzooka_track_id for optimal performance.")


if __name__ == "__main__":
    # Call the main function
    main(session)
```
