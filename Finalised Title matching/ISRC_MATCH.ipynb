```python

--
WITH DISTINCT_TITLE_MATCHES AS (
SELECT DISTINCT 
       APRA_WORK_ID,
       MUZOOKA_TRACK_ID,
FROM ADC_WORKS_TITLE_MATCHES_COMPLETE

)
,base_matching AS (
    SELECT 
        ROW_NUMBER() OVER (ORDER BY i.ADC_ISRC_ID) AS RDC_ISRC_ID,
        i.APRA_WORK_ID,
        c.MUZOOKA_TRACK_ID,
        i.ISRC
    FROM 
    DISTINCT_TITLE_MATCHES c
       INNER JOIN  ADCISRC i
            ON i.APRA_WORK_ID = c.APRA_WORK_ID
)
SELECT * FROM base_matching


```
